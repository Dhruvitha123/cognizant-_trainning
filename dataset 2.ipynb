```python
import pandas as pd
data=pd.DataFrame({
    "name":["ammu","dhruv","ammu","aksh","dhruv"],
    "age":[19,20,19,24,24]})
print(data.duplicated())
```

```python
data1=pd.DataFrame({
    "fruit":["mango","apple","mango","peach","apple"],
    "colour":["yellow","red","yellow","pink","red"]})
print(data1.duplicated())
```

```python
data1=pd.DataFrame({
    "fruit":["mango","apple","mango","peach","apple"],
    "colour":["yellow","red","yellow","pink","red"]})
print(data1.drop_duplicates())
```

```python
employee={
   "name":["ani","ashu","akshu","sweety"],
   "salary":[400000,23000,12777,30050]
}
result=pd.DataFrame(employee)
print(result)
```

```python
employee={
   "name":["nick","ammu","jonas","sweety"],
   "salary":[45000,13000,16777,34050]
}
result=pd.DataFrame(employee)
print(result[result["salary"]>15000])
```

```python
employee={
   "name":["janu","anan","rudin","sweety"],
   "salary":[65000,32000,54707,34050]
}
result=pd.DataFrame(employee)
print(result.query("salary>35000"))
```

```python
employee={
   "name":["janu","anan","rudin","sweety"],
   "salary":[65000,32000,54707,34050]
}
result=pd.DataFrame(employee)
print(result.query("salary>15000 and salary<23069"))
```
