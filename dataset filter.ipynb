```python
import pandas as pd
datajoin=pd.DataFrame({
    "ID":[1,2,3],
    "Name":["lilly","dhruv","sweety"]})
print(datajoin)
    
```

```python
datajoin2=pd.DataFrame({
    "ID":[1,2,3],
    "num":["45","56","67"]})
print(datajoin2)
```

```python
result=pd.merge(datajoin,datajoin2,on="ID")
print(result)
```

```python
pd.merge(datajoin,datajoin2,on="ID",how="inner")
```

```python
import pandas as pd
leftjoin=pd.merge(datajoin,datajoin2,on="ID",how="right")
print(leftjoin)           
```
