```python
try:
    num=int(input("enter the number"))
    print(10/num)
except ZeroDivisionError:
    print("only zero number")
except ValueError:
    print("only integers")
```

```python
try:
    num=float(input("enter the number"))
    print(10/num)
except ZeroDivisionError:
    print("only zero number")
except ValueError:
    print("only decimals")
```
